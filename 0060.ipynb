```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
air = pd.read_csv('AirQuality.csv', sep=';', decimal=',')
```

```python
air.head()
```

```python
air.dropna(axis=1,how='all',inplace=True)
```

```python
air.replace(-200, pd.NA, inplace=True)
```

```python
print(air)
```

```python
air.dropna(axis=0,how='all',inplace=True)
```

```python
print(air)
```

```python
for col in air.columns[2:]:
    air[col] = pd.to_numeric(air[col],errors='coerce')
```

```python
plt.figure(figsize=(10,5))
plt.plot(air['CO(GT)'] [:100])
plt.title('ss')
plt.xlabel('f')
plt.ylabel('t')
plt.show()
```

```python
plt.figure(figsize=(10,5))
plt.hist(air['CO(GT)'] ,bins=22)
plt.title('ss')
plt.xlabel('f')
plt.ylabel('t')
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.boxplot(y=air['CO(GT)'])
plt.title('ss')
plt.show()
```

```python
plt.figure(figsize=(12,8))
sns.heatmap(air.corr(numeric_only=True), annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()
```

```python
sns.pairplot(air[['CO(GT)', 'NO2(GT)', 'C6H6(GT)']].dropna())
plt.show()
```
